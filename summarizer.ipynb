```python
pdf_path = r"C:\Users\sacha\Downloads\hdr2025report.pdf"

try: 
    with pdfplumber.open(pdf_path) as pdf:
        first_page=pdf.pages[0]
        text = first_page.extract_text()

        print("Successfully opened the pdf")
        print(f"Total number of pages:{len(pdf.pages)}")
        print("\n---Here is the text from Page 1---")
        print(text[:500])
except Exception as e:
    print(f"❌ Something went wrong: {e}")
```

```python
file_path = r"C:\Users\sacha\Downloads\hdr2025report.pdf"
text= ""

with pdfplumber.open(file_path) as pdf:
    total= len(pdf.pages)

    for i, page in enumerate(pdf.pages[:50]):
        content = page.extract_text()

        if content:
            text += content
        # Every 50 pages, let us know how it's going
        if (i + 1) % 50 == 0:
            print(f"Done with {i + 1} of {total} pages...")

```

```python
print(f"✅ Finished! Total characters extracted: {len(text)}")
print("\n--- Snippet from the end of the report ---")
print(text[-500:])
```

```python
# Define the persona and the structure
analysis_prompt = f"""
ROLE: You are an AI-powered Document Analysis Tool specialized in UNDP and World Bank reports.
TASK: Process the provided 324-page report and extract structured insights.

FORMAT YOUR RESPONSE AS FOLLOWS:

1. **Executive Overview**: A high-level summary of the report's primary thesis.
2. **Key Policy Metrics**: Extract 3-5 critical data points or rankings found in the text.
3. **Thematic Pillars**: Identify the 3 main pillars of this year's development strategy.
4. **Natural Language Insights**: Answer the following: "What is the most urgent risk identified for developing nations in 2026?"

REPORT DATA:
{text}
"""
```

```python
try:
    #send the request
    response = model.generate_content(analysis_prompt)

    print("\n" + "="*50)
    print("ANALYSIS COMPLETE - STRUCTURED INSIGHTS")
    print("="*50 + "\n")
    print(response.text)

except Exception as e:
    print(f"❌ An error occurred during analysis: {e}")
```

```python
with open("summary_output.txt", "w", encoding="utf-8") as f:
    f.write(response.text)
print("✅ Summary saved to summary_output.txt")
```
